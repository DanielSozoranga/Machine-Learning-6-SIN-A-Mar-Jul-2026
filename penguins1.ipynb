# Visualization

```python
sns.scatterplot(x='culmen_length_mm',y='culmen_depth_mm',data=df,hue='species',palette='Dark2')
```

```python
sns.pairplot(df,hue='species',palette='Dark2')
```

The pairplot shows the relationship between all numeric variables grouped by species.
Each scatter plot compares two variables, while the diagonal shows the distribution of each one.
We can see that Gentoo penguins tend to have longer flippers and heavier bodies compared to Adelie and Chinstrap.
The variables flipper_length_mm and body_mass_g are the most useful to distinguish Gentoo from the other two species.

```python
sns.catplot(x='species',y='culmen_length_mm',data=df,kind='box',col='sex',palette='Dark2')
```

The boxplot compares the culmen length across the three penguin species, separated by sex.
In both males and females, Chinstrap penguins have the longest culmen, followed by Gentoo and then Adelie with the shortest.
Males consistently show longer culmen lengths than females across all species.
Adelie penguins show the most variability and have a few outliers on both ends.

We convert all categorical columns (except species) into numeric dummy variables,
dropping the first category of each to avoid multicollinearity.

```python
X = pd.get_dummies(df.drop('species',axis=1),drop_first=True)
y = df['species']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=967)
```

# Decision Tree Classifier
## Default Hyperparameters

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)

base_pred = model.predict(X_test)
```

## Evaluation


```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, base_pred)
ConfusionMatrixDisplay(cm).plot()
```

```python
print(classification_report(y_test,base_pred))
```

Support is the number of actual instances of each class in the test set.
It tells us how many real samples were used to evaluate each species.

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12,8))
plot_tree(model);
```

```python
plt.figure(figsize=(12,8),dpi=150)
plot_tree(model,filled=True,feature_names=X.columns);
```

Regularization limits the complexity of the decision tree to avoid overfitting.

```python
# Limits the maximum number of levels the tree can grow
DecisionTreeClassifier(max_depth=3)

```

```python
# Sets the minimum number of samples required at a leaf node
DecisionTreeClassifier(min_samples_leaf=10)
```
